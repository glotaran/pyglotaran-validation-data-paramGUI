# simData04 analysis

Notebook analysis for the `simData04` case study.

The code below keeps the workflow explicit so it stays easy to compare with the original R reference.

```python
from glotaran.io import load_dataset, save_result
from glotaran.optimization.optimize import optimize
from glotaran.project.scheme import Scheme
from pyglotaran_extras.io import setup_case_study
from pyglotaran_extras.plotting.plot_overview import plot_overview

base_results_folder, _ = setup_case_study(
    output_folder_name="pyglotaran_validation_param_gui_results"
)

dataset = load_dataset("simData04.ascii")
dataset
```

## Kinetic model

Custom-axis kinetic fit with IRF and fixed initial concentrations.

```python
kin_scheme = Scheme(
    model="simData4/kin_model_ds4.yml",
    parameters="simData4/kin_params_ds4.yml",
    data={"dataset": dataset},
)
kin_result = optimize(kin_scheme)
save_result(kin_result, base_results_folder / "kin/result.yml", allow_overwrite=True)

kin_result
```

```python
fig, _ = plot_overview(kin_result.data["dataset"], linlog=False)

fig.suptitle("simData04 kinetic");
```

## Spectral model

Spectral fit using the skewed-Gaussian basis for the two species.

```python
spectral_scheme = Scheme(
    model="simData4/spectral_model_ds4.yml",
    parameters="simData4/spectral_params_ds4.yml",
    data={"dataset": dataset},
)
spectral_result = optimize(spectral_scheme)
save_result(spectral_result, base_results_folder / "spectral/result.yml", allow_overwrite=True)
spectral_result
```

```python
fig, _ = plot_overview(spectral_result.data["dataset"], linlog=False)

fig.suptitle("simData04 spectral");
```

## Spectrotemporal model

Combined kinetic and spectral fit using a generated full-model definition for this case.

```python
spectemp_scheme = Scheme(
    model="simData4/full_model_ds4.yml",
    parameters="simData4/full_params_ds4.yml",
    data={"dataset": dataset},
)
spectemp_result = optimize(spectemp_scheme)
save_result(
    spectemp_result, base_results_folder / "spectrotemporal/result.yml", allow_overwrite=True
)
spectemp_result
```

```python
fig, _ = plot_overview(spectemp_result.data["dataset"], linlog=True, linthresh=80)

fig.suptitle("simData04 spectrotemporal");
```
